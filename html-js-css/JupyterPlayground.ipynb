```python
class Student:
    def __init__(self, name, student_id, major):
        self.name = name 
        self.student_id = student_id
        self.major = major

    def display_info(self):
        print(f"Student Name: {self.name} Student ID: {self.student_id} Major: {self.major}")

me = Student("Cossette", 10828357, "Computer Science")
me.display_info()
```
